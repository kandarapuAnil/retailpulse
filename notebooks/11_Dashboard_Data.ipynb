```python
import pandas as pd
import numpy as np
```

```python
sales = pd.read_csv("../data/processed/monthly_sales.csv")

segments = pd.read_csv("../data/processed/customer_segments.csv")

forecast = pd.read_csv("../data/processed/hybrid_forecast.csv")

inventory = pd.read_csv("../data/processed/inventory_recommendations.csv")
```

```python
dashboard = {}

dashboard["Total TotalPrice"] = sales["TotalPrice"].sum()

dashboard["Average TotalPrice"] = sales["TotalPrice"].mean()

dashboard["Customers"] = len(segments)

dashboard["Inventory Items"] = len(inventory)

dashboard
```

```python
pd.DataFrame([dashboard]).to_csv(
    "../data/processed/dashboard_kpi.csv",
    index=False
)
```

```python
top_customers = segments.sort_values(
    "Monetary",
    ascending=False
).head(20)

top_customers.to_csv(
    "../data/processed/dashboard_top_customers.csv",
    index=False
)
```

```python
forecast.to_csv(
    "../data/processed/dashboard_forecast.csv",
    index=False
)
```

```python
inventory.to_csv(
    "../data/processed/dashboard_inventory.csv",
    index=False
)
```

```python
import pandas as pd

# Load cleaned data
df = pd.read_csv("../data/processed/cleaned_retail.csv")

# Create TotalPrice if needed
if "TotalPrice" not in df.columns:
    if "Quantity" in df.columns and "Price" in df.columns:
        df["TotalPrice"] = df["Quantity"] * df["Price"]
    elif "Quantity" in df.columns and "UnitPrice" in df.columns:
        df["TotalPrice"] = df["Quantity"] * df["UnitPrice"]

# -----------------------
# Top Products
# -----------------------
products = (
    df.groupby("Description", as_index=False)["TotalPrice"]
      .sum()
      .sort_values("TotalPrice", ascending=False)
)

products.to_csv(
    "../data/processed/dashboard_products.csv",
    index=False
)

# -----------------------
# Country Sales
# -----------------------
country = (
    df.groupby("Country", as_index=False)["TotalPrice"]
      .sum()
      .sort_values("TotalPrice", ascending=False)
)

country.to_csv(
    "../data/processed/dashboard_country.csv",
    index=False
)

print("✅ dashboard_products.csv created")
print("✅ dashboard_country.csv created")
```
